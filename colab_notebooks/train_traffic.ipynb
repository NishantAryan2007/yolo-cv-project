## Step 1: Download the dataset from Roboflow

```python
!pip install roboflow

from roboflow import Roboflow
rf = Roboflow(api_key="YOUR_API_KEY_HERE")
project = rf.workspace("learning-evidence").project("helmet-detection_yolov8")
version = project.version(3)
dataset = version.download("yolov8")

```

## Step 2: Verify the dataset structure and class list

Confirming the class names, order, and folder paths Roboflow generated,
before starting training.

```python
import os

for root, dirs, files in os.walk("."):
    for file in files:
        if file == "data.yaml":
            path = os.path.join(root, file)
            print("Found:", path)
            with open(path) as f:
                print(f.read())
```

## Step 3: Train the YOLOv8 model



```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data = "Helmet-Detection_YOLOv8-3/data.yaml",
    epochs = 50,
    imgsz = 640,
    batch = 16,
    project = "traffic_training_v2",
    name = "helmet_run1"
)
```
